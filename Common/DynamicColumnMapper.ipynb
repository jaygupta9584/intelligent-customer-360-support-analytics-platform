```python

import re
from collections import Counter


def normalize_column_name(column_name):
    name = column_name.strip()

    # CustomerId -> Customer_Id
    name = re.sub(
        r"([a-z0-9])([A-Z])",
        r"\1_\2",
        name
    )

    # Lowercase
    name = name.lower()

    # Replace spaces and special characters
    name = re.sub(r"[^a-z0-9]+", "_", name)

    return name.strip("_")

def dynamic_column_mapping(df):
    original_columns = df.columns

    new_columns = [
        normalize_column_name(col)
        for col in original_columns
    ]

    if any(not col for col in new_columns):
        raise ValueError(
            "Empty column name after normalization"
        )

    duplicates = [
        name
        for name, count in Counter(new_columns).items()
        if count > 1
    ]

    if duplicates:
        raise ValueError(
            f"Column collisions detected: {duplicates}"
        )

    for old, new in zip(original_columns, new_columns):
        if old != new:
            df = df.withColumnRenamed(old, new)

    return df
```
