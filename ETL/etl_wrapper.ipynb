# ETL Wrapper

Runs bronze → silver for specified tables via `dbutils.notebook.run()`.  
**Parameter:** `tables` — comma-separated table names (e.g. `customers,orders`).

```python
# ETL Wrapper: runs bronze then silver for each table in the `tables` parameter
# Called by the Customer 360 ETL Pipeline job with different table groups.

base_bronze = "/Users/jaygupta9584@gmail.com/Intelligent Customer 360 & Support Analytics Platform/Bronze/"
base_silver = "/Users/jaygupta9584@gmail.com/Intelligent Customer 360 & Support Analytics Platform/Silver/"

table_map = {
    "customers":       ("bronze_customers.ipynb",      "silver_customers"),
    "orders":          ("bronze_orders.ipynb",         "silver_orders"),
    "payments":        ("bronze_payments.ipynb",       "silver_payments"),
    "products":        ("bronze_products.ipynb",       "silver_products"),
    "order_items":     ("bronze_order_items.ipynb",    "silver_order_items"),
    "support_tickets": ("bronze_support_tickets.ipynb","silver_support_tickets"),
    "customer_events": ("bronze_customer_events.ipynb","silver_customer_events"),
}

dbutils.widgets.text("tables", "")
tables = [t.strip() for t in dbutils.widgets.get("tables").split(",") if t.strip()]

print(f"ETL Wrapper: processing {len(tables)} table(s): {', '.join(tables)}")
print("=" * 60)

for table in tables:
    bronze_file, silver_notebook = table_map[table]
    bronze_path = base_bronze + bronze_file
    silver_path = base_silver + silver_notebook

    print(f"\n--- [{table}] Running bronze: {bronze_file} ---")
    dbutils.notebook.run(bronze_path, 1200)

    print(f"--- [{table}] Running silver: {silver_notebook} ---")
    dbutils.notebook.run(silver_path, 1200)

    print(f"=== [{table}] Completed ===")

print("\n" + "=" * 60)
print(f"All {len(tables)} table(s) processed successfully: {', '.join(tables)}")
```
