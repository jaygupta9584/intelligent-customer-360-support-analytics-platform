```python
bronze_table_name = f"{catalog_name}.{bronze_schema}.support_tickets"
source_volumne = f"{landing_folder_path}support_tickets.csv"
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType

support_tickets_schema = StructType([
    StructField("ticket_id", IntegerType(), True),
    StructField("customer_id", IntegerType(), True),
    StructField("order_id", IntegerType(), True),
    StructField("issue_category", StringType(), True),
    StructField("subject", StringType(), True),
    StructField("description", StringType(), True),
    StructField("priority", StringType(), True),
    StructField("status", StringType(), True),
    StructField("resolution", StringType(), True),
    StructField("created_at", TimestampType(), True),
    StructField("resolved_at", TimestampType(), True),
    StructField("updated_at", TimestampType(), True)
])
```

```python
df = spark.read.csv(source_volumne, header=True, schema=support_tickets_schema)

finaldf = add_ingestion_metadata(df, source_volumne)

load_to_bronze(bronze_table_name, source_volumne, finaldf)

display(spark.read.table(bronze_table_name))
```
