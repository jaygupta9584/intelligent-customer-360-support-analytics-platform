```python
bronze_table_name = f"{catalog_name}.{bronze_schema}.products"
source_volumne = f"{landing_folder_path}products.csv"
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType, TimestampType

product_schema = StructType([
    StructField("product_id", IntegerType(), True),
    StructField("product_name", StringType(), True),
    StructField("category", StringType(), True),
    StructField("sub_category", StringType(), True),
    StructField("brand", StringType(), True),
    StructField("price", DoubleType(), True),
    StructField("cost", DoubleType(), True),
    StructField("created_at", TimestampType(), True),
    StructField("updated_at", TimestampType(), True)
])
```

```python
df = spark.read.csv(source_volumne, header=True, schema=product_schema)

finaldf = add_ingestion_metadata(df, source_volumne)

try:
    load_to_bronze(bronze_table_name, source_volumne, finaldf)
except Exception as e:
    print(e)


# display(spark.read.table(bronze_table_name))
```
