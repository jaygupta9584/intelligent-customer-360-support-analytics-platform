```python
bronze_table_name = f"{catalog_name}.{bronze_schema}.customers"
source_volumne = f"{landing_folder_path}customers.csv"
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, TimestampType

customers_schema = StructType([
  StructField("id", IntegerType(), True),
  StructField("name", StringType(), True),
  StructField("email", StringType(), True),
  StructField("phone", StringType(), True),
  StructField("address", StringType(), True),
  StructField("city", StringType(), True),
  StructField("state", StringType(), True),
  StructField("zip", StringType(), True),
  StructField("country", StringType(), True),
  StructField("created_at", TimestampType(), True),
  StructField("updated_at", TimestampType(), True)
])
```

```python
# Read CSV with explicit schema
df = spark.read.csv(source_volumne, header=True, schema=customers_schema)

# Add ingestion metadata
finaldf = add_ingestion_metadata(df, source_volumne)

try:
    load_to_bronze(bronze_table_name, source_volumne, finaldf)
except Exception as e:
    print(e)


display(spark.read.table(bronze_table_name))
```
